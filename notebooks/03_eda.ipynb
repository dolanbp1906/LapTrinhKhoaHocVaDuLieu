```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path('..').resolve()
PROC = ROOT / 'data' / 'processed'
FIG = ROOT / 'reports' / 'figures'
sns.set_theme(style='whitegrid')

fact = pd.read_csv(PROC / 'fact_sales.csv', parse_dates=['order_date'])
stock = pd.read_csv(PROC / 'summaries' / 'stock_vs_reorder.csv')
rfm = pd.read_csv(PROC / 'rfm_raw.csv')
print(fact.shape, stock.shape, rfm.shape)
fact.head()
```

## Doanh thu theo tháng & danh mục

```python
by_month = fact.groupby(fact['order_date'].dt.to_period('M').astype(str))['net_revenue'].sum()
by_cat = fact.groupby('category')['net_revenue'].sum().sort_values(ascending=False)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
by_month.plot(marker='o', ax=ax[0], title='Doanh thu theo tháng')
ax[0].tick_params(axis='x', rotation=45)
by_cat.head(6).plot(kind='barh', ax=ax[1], title='Top danh mục')
ax[1].invert_yaxis()
plt.tight_layout()
plt.show()
```

## Giá trị đơn & kênh bán

```python
order_val = fact.groupby('order_id')['net_revenue'].sum()
print(order_val.describe())
fact.groupby('sales_channel')['net_revenue'].sum().sort_values(ascending=False)
```

## Tồn kho vs reorder & RFM thô

```python
print('Below reorder:', int(stock['below_reorder'].sum()), '/', len(stock))
display(rfm.describe())
```
